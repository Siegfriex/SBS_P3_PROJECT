# P3 Culture Raw Data EDA

`P3_DATA_RAW` 브랜치의 원본 엑셀 구조를 확인하는 스타터 노트북입니다. 원본 파일은 수정하지 않고, 시트/컬럼/샘플 행을 먼저 점검합니다.

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd()
DATA_ORIGIN = PROJECT_ROOT / "data_origin"
RAW_XLSX = DATA_ORIGIN / "국정감사회의록_문화체육 2020~.xlsx"

RAW_XLSX.exists(), RAW_XLSX
```

```python
xlsx = pd.ExcelFile(RAW_XLSX)
xlsx.sheet_names
```

```python
sheet_overview = []

for sheet_name in xlsx.sheet_names:
    preview = pd.read_excel(RAW_XLSX, sheet_name=sheet_name, nrows=5)
    sheet_overview.append(
        {
            "sheet_name": sheet_name,
            "n_preview_rows": len(preview),
            "n_columns": len(preview.columns),
            "columns": list(preview.columns),
        }
    )

pd.DataFrame(sheet_overview)
```

```python
first_sheet = xlsx.sheet_names[0]
raw_preview = pd.read_excel(RAW_XLSX, sheet_name=first_sheet, nrows=20)
raw_preview
```
